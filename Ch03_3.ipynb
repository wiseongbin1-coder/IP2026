```python
def saturation(value):
  if(value > 255):
    value = 255
  elif(value < 0):
    value = 0
  return value
```

```python
import cv2

img = cv2.imread('/content/gdrive/My Drive/ColabIP26/Images/aero2.bmp')
gimg = cv2.cvtColor(img, cv2.COLOR_BGR2GRAY)

cv2_imshow(gimg)
```

```python
gOut = np.zeros((gimg.shape[0], gimg.shape[1]), dtype=np.ubyte)
```

```python
min = 256
max = 0

for h in range(gimg.shape[0]):
 for w in range(gimg.shape[1]):
  if gimg[h, w] < min:
    min = gimg[h, w]
  if gimg[h, w] >max:
    max = gimg[h,w]
print(min)
print(max)
```

```python
for h in range(gimg.shape[0]):
 for w in range(gimg.shape[1]):
  val = (gimg[h, w] - min) / (max - min) * 255
  gOut[h, w] = val

cv2_imshow(gOut)
```

```python
min = 256
max = 0

for h in range(gimg.shape[0]):
 for w in range(gimg.shape[1]):
  if gOut[h, w] < min:
    min = gOut[h, w]
  if gOut[h, w] >max:
    max = gOut[h,w]

print(min)
print(max)
```

```python
img = cv2.imread('/content/gdrive/My Drive/ColabIP26/Images/girl.jpg')
gimg1 = cv2.cvtColor(img, cv2.COLOR_BGR2GRAY)

cv2_imshow(gimg1)
```

```python
gMul = np.zeros((gimg.shape[0], gimg.shape[1]), dtype=np.ubyte)
```

```python
mulVal = 2

for h in range(gimg.shape[0]):
 for w in range(gimg.shape[1]):
  val = np.int32(gimg1[h, w]) * mulVal
  gMul[h, w] = val

cv2_imshow(gMul)
```

```python
gDiv = np.zeros((gimg.shape[0], gimg.shape[1]), dtype=np.ubyte)
```

```python
divVal = 4

for h in range(gimg.shape[0]):
 for w in range(gimg.shape[1]):
  val = np.int32(gMul[h, w]) / divVal
  gMul[h, w] = saturation(val)

cv2_imshow(gDiv)
```
