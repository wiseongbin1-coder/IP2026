```python
import cv2
from google.colab.patches import cv2_imshow
import numpy as np
import matplotlib.pyplot as plt
from google.colab import drive
drive.mount('/content/gdrive')
```

```python
img = cv2.imread('/content/gdrive/My Drive/ColabIP26/Images/lena256.jpg')
RGB_img = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)

gimg1 = cv2.cvtColor(img, cv2.COLOR_BGR2GRAY)
print(gimg1.shape)
cv2_imshow(gimg1)
```

```python
hist = cv2.calcHist([gimg1],[0],None,[256],[0,256])
plt.plot(hist)
```

```python
gOut = cv2.equalizeHist(gimg1)
cv2_imshow(gOut)
```

```python
hist2 = cv2.calcHist([gOut],[0],None,[256],[0,256])
plt.plot(hist2)
```

```python
cumuHist = hist.copy()
for i in range(1,len(cumuHist)):
  cumuHist[i] = cumuHist[i-1] + hist2[i]

plt.plot(cumuHist)
```
