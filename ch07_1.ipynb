```python
img = cv2.imread('/content/gdrive/My Drive/ColabIP26/Images/lena256.jpg')
RGB_img = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)

gimg = cv2.cvtColor(RGB_img, cv2.COLOR_BGR2GRAY)
print(gimg.shape)
cv2_imshow(gimg)
```

```python
height, width = gimg.shape
print(height, width)
```

```python
R=cv2.getRotationMatrix2D((0,0), 45, 1)
#R=cv2.getRotationMatrix2D(((width-1)/2, (height-1)/2), 45, 1)
outimg = cv2.warpAffine(gimg, R, (width, height))
cv2_imshow(outimg)
```
