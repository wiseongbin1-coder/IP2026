```python
import cv2
from google.colab.patches import cv2_imshow
import numpy as np
import matplotlib.pyplot as plt
from google.colab import drive
drive.mount('/content/gdrive')
```

```python
img = cv2.imread('/content/gdrive/My Drive/ColabIP26/Images/lena256.jpg')
RGB_img = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)

gimg1 = cv2.cvtColor(img, cv2.COLOR_BGR2GRAY)
print(gimg1.shape)
cv2_imshow(gimg1)
```

```python
hist = cv2.calcHist([gimg1],[0],None,[256],[0,255])
hist = cv2.calcHist([RGB_img],[2],None,[256],[0,255])
plt.plot(hist)
```

```python
gBin = np.zeros((gimg1.shape[0], gimg1.shape[1]), np.uint8)
```

```python
T = 180

for h in range(gimg1.shape[0]):
  for w in range(gimg1.shape[1]):
    if gimg1[h, w] > T:
      gBin[h, w] = 255
    else:
      gBin[h, w] = 0

cv2_imshow(gBin)
```
