```python
import cv2
from google.colab.patches import cv2_imshow
import numpy as np
import matplotlib.pyplot as plt
from google.colab import drive
drive.mount('/content/gdrive')
```

```python
img = cv2.imread('/content/gdrive/My Drive/ColabIP26/Images/scanimg.jpg')
RGB_img = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)
gimg = cv2.cvtColor(RGB_img, cv2.COLOR_BGR2GRAY)
print(gimg.shape)
cv2_imshow(gimg)
```

```python
kernel = np.ones((4,4),np.uint8)
dilated = cv2.dilate(gimg,kernel,iterations = 1)
cv2_imshow(dilated)
```

```python
kernel = np.ones((4,4),np.uint8)
eroded = cv2.erode(dilated,kernel,iterations = 1)
cv2_imshow(eroded)
```

```python
img = cv2.imread('/content/gdrive/My Drive/ColabIP26/Images/rice.bmp')
RGB_img = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)
gimg = cv2.cvtColor(RGB_img, cv2.COLOR_BGR2GRAY)
print(gimg.shape)
cv2_imshow(gimg)
```

```python
gBin = np.zeros((gimg.shape[0],gimg.shape[1]))
```

```python
threshold = 120

for i in range(gimg.shape[0]):
  for j in range(gimg.shape[1]):
    if gimg[i,j] > threshold:
      gBin[i,j] = 255

cv2_imshow(gBin)
```

```python
kernel = np.ones((3,3),np.uint8)
eroded = cv2.erode(gBin,kernel,iterations = 1)
cv2_imshow(eroded)
```

```python
kernel = np.ones((3,3),np.uint8)
openimg = cv2.morphologyEx(gBin, cv2.MORPH_OPEN, kernel)
cv2_imshow(openimg)
```

```python
kernel = np.ones((3,3),np.uint8)
dilated = cv2.dilate(eroded,kernel,iterations = 1)
cv2_imshow(dilated)
```

```python
eroded = eroded.astype(np.uint8)
```
