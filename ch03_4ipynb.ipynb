```python
import cv2
from google.colab.patches import cv2_imshow
import numpy as np
import matplotlib.pyplot as plt
from google.colab import drive
drive.mount('/content/gdrive')
```

```python
def saturation(value):
  if(value > 255):
    value = 255
  elif(value < 0):
    value = 0
  return value
```

```python
img = cv2.imread('/content/gdrive/My Drive/ColabIP26/Images/lena256.jpg')
gimg1 = cv2.cvtColor(img, cv2.COLOR_BGR2GRAY)

cv2_imshow(gimg1)
```

```python
gOut = np.zeros((gimg1.shape[0], gimg1.shape[1]), dtype=np.ubyte)
```

```python
planeNumber = 8
planeValue = 1

for i in range(planeNumber-1):
  planeValue = planeValue * 2
```

```python
for h in range(gimg1.shape[0]):
 for w in range(gimg1.shape[1]):
  imVal = np.int32(gimg1[h,w] / planeValue);
  if(imVal % 2 == 0):
    gOut[h,w] = 0
  else:
    gOut[h,w] = 255

cv2_imshow(gOut)
```
